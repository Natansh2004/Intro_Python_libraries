```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('Used_Bikes.csv')
df.head()
```

```python
# Removing all the duplicate data permanently
df.drop_duplicates(inplace=True)
```

```python
# we need to make a bar plot that shows bike brands 
# and number of bikes in each brand
```

```python
# counting the total number of bikes present in each brand
df['brand'].value_counts()
```

```python
bikes = list(df['brand'].value_counts().keys())
bikes
```

```python
bikes_no = list(df['brand'].value_counts().values)
bikes_no
```

```python
plt.bar(bikes,bikes_no)
plt.xlabel('Bike Brands')
plt.ylabel('No. of bikes')
plt.show()

# but we can see here that on x-axis, names are overlapping
# there are 2 solutions to this
#   1) increasing figure size
#   2) by rotating x ticks
```

```python
##   1st SOLUTION FOR OVERLAPPING :---->

# Customize size according to you by -> plt.figure(figsize=(15,8))
# figsize = (15,8)     
# Here 15 is width in inches
# And 8 is height in inches
# figsize = (width, height)

plt.figure(figsize=(32,10))   
plt.bar(bikes,bikes_no,color='green',ec='k',lw=1)
# ec = edge color... ec = 'k'.. k represents black 
# (it is shortcut for black color)
# lw = line width
plt.xlabel('Brands')
plt.ylabel('No. of bikes')
plt.title('Chart on brand column')   # Giving title to barchart
plt.show()
```

```python
##  2nd SOLUTION FOR OVERLAPPING  (BETTER SOLUTION):---->

# By rotating x-ticks

plt.bar(bikes,bikes_no,color='green',ec='k',lw=1)
plt.xlabel('Brands')
plt.ylabel('No. of bikes')
plt.title('Chart on brand column')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(15,8))
plt.bar(bikes,bikes_no,color='green',ec='k',lw=1)
plt.xlabel('Brands')
plt.ylabel('No. of bikes')
plt.title('Chart on brand column')
plt.xticks(rotation=60)
plt.show()
```

```python
# if we want we can change line style
# ls = line style

plt.figure(figsize=(15,8))
# customize size according to you
plt.bar(bikes,bikes_no,color='orange',ec='k',lw=2,ls='dashed')
#ls='dashed'
# for black-->k,ec=edge color,lw=line width,ls=line style
plt.ylabel("no of bikes")
plt.xlabel("brands")
plt.title("Chart on brand column")
plt.xticks(rotation=60)
plt.show()
```

```python
plt.figure(figsize=(15,8))
plt.bar(bikes,bikes_no,color='orange',ec='k',lw=2,ls='dotted')
#ls='dotted'
plt.ylabel("no of bikes")
plt.xlabel("brands")
plt.title("Chart on brand column")
plt.xticks(rotation=60)
plt.show()
```

```python
##     PIE PLOT

# A pie plot, also known as a pie chart or a circle graph, is a 
# graphical representation of data where each slice of the pie 
# represents a proportion of the whole. It is commonly used to 
# visualize categorical data, where each category is represented 
# by a different color or pattern.

print(country_name)
print(country_population)
```

```python
# 'labels' should have categorical value
plt.pie(country_population,labels=country_name)
plt.show()
```

```python
# To calculate percentage, we use autopct
# autopct = auto percentage

plt.pie(country_population,labels=country_name,autopct="%2.2f%%")
# .4etc---->decimal value upto 4 point according to you
# calculate auto percentage
plt.show()
```

```python
# customising size of the pie chart

plt.figure(figsize=(8,8))
plt.pie(country_population,labels=country_name,autopct="%2.2f%%")
plt.show()
```

```python
plt.figure(figsize=(8,8))
plt.pie(country_population,labels=country_name,autopct="%2.2f%%",
        wedgeprops={'ec':"black",'ls':'dashed','lw':2})
plt.show()

# wedgeprops is a dictionary    {'key':'value'} pair
# ec = edge color      
# ls = line style
# lw = line width
```

```python
# explode list --> used to apply some distance from the main chart 
# --> used to highlight that particular part...

print(country_name)
explosion = [0,0,0,0.2]
#plt.figure(figsize=(8,8))
plt.pie(country_population,labels=country_name,autopct='%2.2f%%', \
    wedgeprops={'ec':'black','ls':'dashed','lw':2},explode=explosion)
plt.show()
```

```python
# to apply shadow to the pie chart

print(country_name)
explosion = [0,0.2,0,0]
#plt.figure(figsize=(8,8))
plt.pie(country_population,labels=country_name,autopct='%2.2f%%', \
    wedgeprops={'ec':'black','lw':2},explode=explosion,shadow=True)
plt.show()
```

```python
# saving any chart

plt.pie(country_population,labels=country_name,autopct='%2.2f%%',\
        wedgeprops={'ec':'k','lw':2})
plt.savefig('Pie_chart image')
plt.show()
```

```python
# Histogram: It tells how our data is distributed. A histogram plot 
#     is a graphical representation of data where the data is divided
#     into bins and the frequency of each bin is represented by a bar.
#     It is commonly used to visualize the distribution of data and 
#     identify patterns,trends,and outliers.

#  Here are some key points about histogram plots:
#   1) Each bar in the histogram represents a bin, which is a range 
#      of values.
#   2) The height of each bar represents the frequency or count of 
#      data points that fall within that bin.
#   3) The width of each bar can be adjusted to represent the size
#      of the bins.
#   4) Histograms can be used to compare the distribution of data 
#      between different groups or categories.
#   5) Histograms can be customized by changing the color, style, 
#      and other visual properties.
#   6) Histograms can be used to identify patterns, trends, and 
#      outliers in the data.
```
