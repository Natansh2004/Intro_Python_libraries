```python
# The sns.barplot function in seaborn is used to create a bar plot, 
# which is a type of plot that displays the relationship between a 
# categorical variable and a numerical variable. It takes the 
# following parameters: 
#     x: The data for the x-axis (the categorical variable).
#     y: The data for the y-axis (the numerical variable).
```

```python
sns.barplot(x='sex',y='total_bill',data=df)
plt.show()
```

```python
sns.histplot(x='sex',y='tip',data=df)
plt.show()
```

```python
# Seaborn has two important features: relplot and catplot.

# 1.relplot: This function creates a figure-level interface for 
# drawing relational plots. It is used to visualize the relationship 
# between two numerical variables. By default, it creates scatter 
# plots, but it can also create line plots by setting the kind 
# parameter.

# 2.catplot: This function creates a figure-level interface for 
# drawing categorical plots. It is used to visualize the 
# relationship between a categorical variable and a numerical 
# variable. It can create various types of plots, such as 
# stripplots, swarmplots, boxplots, violinplots, barplots, 
# and point plots, by setting the kind parameter.

# seaborn has 2 important function---- 
#     1.relplot()-->if both row and column are numerical 
#     2.catplot()-->if one is numerical and other is categorical

#(relplot)relational plot by default create the scatter plot
```

```python
sns.relplot(x='tip',y='total_bill',data=df)
plt.show()
```

```python
sns.relplot(x='tip',y='total_bill',data=df,hue='sex')
plt.show()
```

```python
sns.relplot(x='tip',y='total_bill',data=df,hue='sex',style='smoker')
plt.show()
```

```python
sns.relplot(x='tip',y='total_bill',data=df,hue='sex',style='smoker',col='day')
plt.show()
```

```python
sns.relplot(x='tip',y='total_bill',data=df,hue='time',style='smoker',col='day')
plt.show()

# This ORANGE in the thursday is the OUTLIER
```

```python
sns.kdeplot(df['total_bill'])
plt.show()

# normal distribution, so we'll find mean in the pandas for 
# the null values.

# About Kdeplot: sns.kdeplot() function in seaborn is used 
# to create a kernel density estimate (KDE) plot. It 
# provides a flexible interface for visualizing the 
# distribution of a numerical variable.
```

```python
sns.displot(df['total_bill'],kde=True,color='g')
plt.title('Histogram with kde')
plt.show()
# histogram 
```

```python
sns.kdeplot(df['size'])
plt.show()
# histogram 

# not a normal distribution, so we'll find median in the pandas 
# for the null values.
```
