```python
## DATA CLEANING
#  MISSING VALUE HANDLING
import numpy as np
import pandas as pd
```

```python
data = {'A':[10,np.nan,20,54,30,40,np.nan],
        'B':[np.nan,50,60,70,np.nan,25,74], 
        'C':[99,50,np.nan,70,np.nan,90,74], 
        'D':[1,2,3,4,5,6,7]}

# np.nan => NaN --- Not a Number  (Missing value)
data 
```

```python
df2 = pd.DataFrame(data)
df2
```

```python
# to find out if there are any null (or NaN) values in the dataframe
df2.isnull()
```

```python
# to find out how many null (or NaN) values are there in each column

df2.isnull().sum()
# column wise NaN values
```

```python
# to find total NaN values in the dataframe
df2.isnull().sum().sum()
```

```python
df2
```

```python
df2.dropna()     #inplace = True

#dropna() method by default removes the rows that contains NULL values.
```

```python
df2.dropna(axis=1)

#axis = 1  (column)
#axis = 0  (rows)
```

```python
df2
```

```python
df2.fillna('Upflairs')

# it will fill the NaN values with the scalar value 'Upflairs'...
# it will replace NaN value in the whole dataframe..
```

```python
# if we want to replace NaN values with some value in a particular 
# column...

df2['B'].fillna('Upflairs')    #inplace = True
```

```python
# no changes made in the original dataframe
df2
```

```python
# to calculate mean in a particular column

df2['C'].mean()

# (99+50+70+90+74)/5 = 383/5 = 76.6
# NaN values will be skipped
```

```python
df2['C'].median()
```

```python
# now filling median value i.e., 74.0 in place of NaN values 
# in column 'C'. 

df2['C'].fillna(df2['C'].median())
```
